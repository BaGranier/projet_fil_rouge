```python
import pandas as pd
import math
import matplotlib.pyplot as plt 
import seaborn as sns
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

filename = "Donnees_projet_fil_rouge.csv"
df = pd.read_csv(filename, header=0, sep=';')  # essayez d'abord avec point-virgule

print(f"Dataset Shape: {df.shape}")

print(df.columns)
```

```python
print(df.head())
```

```python



cat_cols = df.select_dtypes(include=['object']).columns.tolist()
num_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
all_cols = cat_cols + num_cols

if len(all_cols) == 0:
    print("Aucune colonne à afficher.")
else:
    ncols = 3
    ncols = min(ncols, len(all_cols))
    nrows = math.ceil(len(all_cols) / ncols)

    fig, axs = plt.subplots(nrows=nrows, ncols=ncols, figsize=(ncols*5, nrows*4))
    axs = axs.flatten()

    possible_hues = ['cereal', 'Cereal', 'class', 'Class', 'is_cereal']
    hue_col = next((h for h in possible_hues if h in df.columns), None)

    for i, column in enumerate(all_cols):
        ax = axs[i]
        if column in cat_cols:
            if hue_col is not None and hue_col != column:
                sns.countplot(data=df, x=column, hue=hue_col, palette='Set2', ax=ax)
                ax.legend(title=hue_col, loc='upper right', fontsize='small')
            else:
                sns.countplot(data=df, x=column, palette='Set2', ax=ax)
            ax.set_title(f'Repartition of {column}')
        else:
            # histogramme principal
            sns.histplot(data=df, x=column, kde=True, palette='Set2', bins=30, ax=ax)
            ax.set_title(f'Distribution of {column}')

            # inset: petite boxplot sous l'histogramme
            # ajuster le dernier nombre pour augmenter/diminuer la hauteur
            inset_ax = ax.inset_axes([0, -0.30, 1, 0.25])  # [left, bottom, width, height] en fraction de ax
            if hue_col is not None:
                # boxplot groupée par la colonne hue (affiche distribution par classe)
                sns.boxplot(x=hue_col, y=column, data=df, palette='Set2', ax=inset_ax)
                inset_ax.set_xlabel('')
                inset_ax.set_ylabel('')
            else:
                # boxplot simple (horizontale)
                sns.boxplot(x=column, data=df, orient='h', color='lightgrey', ax=inset_ax)
                inset_ax.set_xlabel('')
                inset_ax.set_yticks([])

            # lisibilité: réduire les labels et la taille de police
            inset_ax.tick_params(axis='x', rotation=0, labelsize='small')
            inset_ax.tick_params(axis='y', labelsize='small')

        ax.tick_params(axis='x', rotation=45)

    for j in range(len(all_cols), len(axs)):
        axs[j].set_visible(False)

    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import xgboost as xgb
from sklearn.model_selection import train_test_split

# ---------------------------
# 1. Définition des cibles
# ---------------------------

targets = {
    'EM porc croissance' : 'EM porc croissance (kcal) kcal/kg brut',
    'ED porc croissance' : 'ED porc croissance (kcal) kcal/kg brut',
    'EN porc croissance' : 'EN porc croissance (kcal) kcal/kg brut',
    'EMAn coq'           : 'EMAn coq (kcal) kcal/kg brut',
    'EMAn poulet'        : 'EMAn poulet (kcal) kcal/kg brut',
    'UFL'                : 'UFL 2018 par kg brut',
    'PDI'                : 'PDI 2018 g/kg brut',
    'BalProRu'           : 'BalProRu 2018 g/kg brut'
}

# ---------------------------
# 2. Données explicatives X
# ---------------------------
X = X_mod.copy()

# ---------------------------
# 3. Split commun pour cohérence
# ---------------------------
X_train, X_test = train_test_split(X, test_size=0.2, random_state=62)

# ---------------------------
# 4. Entraînement + importances
# ---------------------------

feature_importances_all = {}

for label, col in targets.items():
    print(f"\n=== Entraînement modèle pour {label} ===")

    y = df[col]
    
    y_train = y.loc[X_train.index]
    y_test  = y.loc[X_test.index]

    model = xgb.XGBRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=6,
        random_state=42
    )

    model.fit(X_train, y_train)

    # Importances
    importances = pd.Series(
        model.feature_importances_,
        index=X.columns
    ).sort_values(ascending=False)

    feature_importances_all[label] = importances


# ---------------------------
# 5. Visualisation : 8 graphes
# ---------------------------

fig, axes = plt.subplots(4, 2, figsize=(16, 20))
axes = axes.flatten()

for i, (label, importances) in enumerate(feature_importances_all.items()):

    top = importances.head(15)

    sns.barplot(
        x=top.values,
        y=top.index,
        palette="viridis",
        ax=axes[i]
    )
    axes[i].set_title(f"Top variables importantes — {label}")
    axes[i].set_xlabel("Importance")
    axes[i].set_ylabel("Variables")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import r2_score
import xgboost as xgb
import matplotlib.pyplot as plt

# ==============================
# 0. Paramètres
# ==============================

sous_classe_col = 'Nom'
y_em = df['EM porc croissance (kcal) kcal/kg brut']
X = X_mod.copy()

# ==============================
# 1. Trouver le nombre minimal de points par sous-classe
# ==============================

counts = df[sous_classe_col].value_counts()
print("Effectifs par sous-classe :")
print(counts)

min_count = counts.min()
print(f"\nNombre minimal de points parmi les sous-classes : {min_count}")

# ==============================
# 2. Boucle avec réduction + CV 10 folds
# ==============================

results = []  # stockera : n_points, mean_R2, std_R2

for n in range(min_count, 2, -10):  # de min_count vers 3

    print(f"\n=== Test avec {n} points / sous-classe ===")

    # Sous-échantillonnage équilibré
    df_bal = (
        df.groupby(sous_classe_col, group_keys=False)
          .apply(lambda g: g.sample(n=n, random_state=42))
    )
    idx = df_bal.index

    X_n = X.loc[idx]
    y_n = y_em.loc[idx]

    # ----------------------------
    # VALIDATION CROISÉE 10 FOLDS
    # ----------------------------
    kf = KFold(n_splits=10, shuffle=True, random_state=42)

    r2_scores = []

    for train_idx, test_idx in kf.split(X_n):
        X_train, X_test = X_n.iloc[train_idx], X_n.iloc[test_idx]
        y_train, y_test = y_n.iloc[train_idx], y_n.iloc[test_idx]

        # Modèle XGBoost
        model = xgb.XGBRegressor(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=6,
            random_state=42
        )

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        r2_scores.append(r2_score(y_test, y_pred))

    mean_r2 = np.mean(r2_scores)
    std_r2 = np.std(r2_scores)

    print(f"R² moyen = {mean_r2:.3f} ± {std_r2:.3f}")

    results.append((n, mean_r2, std_r2))

# ==============================
# 3. Graphique avec barres d’erreurs
# ==============================

results_df = pd.DataFrame(results, columns=['points_par_sous_classe', 'R2_mean', 'R2_std'])

plt.figure(figsize=(10, 6))
plt.errorbar(
    results_df['points_par_sous_classe'],
    results_df['R2_mean'],
    yerr=results_df['R2_std'],
    fmt='-o',
    capsize=4,
    label="R² (moyenne ± écart-type, CV 10 folds)"
)

plt.gca().invert_xaxis()  # on voit la chute du R² en réduisant les points
plt.xlabel('Nombre de points par sous-classe')
plt.ylabel('R² (EM porc croissance)')
plt.title("Impact de l'échantillonnage équilibré (CV 10-fold) sur R²")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

results_df
```
